Bài 4 — Vẽ và so sánh hai đường ROC bằng RocCurveDisplay

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import RocCurveDisplay, roc_auc_score

X, y = load_breast_cancer(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=0)
```

```python
# 1: huấn luyện hai mô hình trên train
model_lr = LogisticRegression(max_iter=5000, random_state=0)
model_lr.fit(X_train, y_train)

model_dt = DecisionTreeClassifier(max_depth=3, random_state=0)
model_dt.fit(X_train, y_train)
```

```python
# 2: vẽ cả hai đường ROC trên cùng 1 trục
fig, ax = plt.subplots(figsize=(7, 7))

RocCurveDisplay.from_estimator(model_lr, X_test, y_test, ax=ax, name="LogisticRegression")
RocCurveDisplay.from_estimator(model_dt, X_test, y_test, ax=ax, name="DecisionTree (depth=3)")

ax.plot([0, 1], [0, 1], linestyle=":", color="gray", label="Random (AUC = 0.5)")
ax.set_title("So sánh ROC: LogisticRegression vs DecisionTree")
ax.legend(loc="lower right")
plt.tight_layout()
plt.show()
plt.close()
```

```python
# 3: tính lại AUC bằng roc_auc_score để xác nhận khớp với legend
auc_lr = roc_auc_score(y_test, model_lr.predict_proba(X_test)[:, 1])
auc_dt = roc_auc_score(y_test, model_dt.predict_proba(X_test)[:, 1])

print("AUC LogisticRegression :", auc_lr)
print("AUC DecisionTree        :", auc_dt)
```
