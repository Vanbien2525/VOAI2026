Bài 3 — Tái hiện cái bẫy labels trên một fold thật

```python
# %%
from sklearn.datasets import make_classification
from sklearn.model_selection import KFold, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
```

```python
X, y = make_classification(n_samples=100, weights=[0.97, 0.03], n_informative=3, random_state=7)
print("Số mẫu lớp thiểu số:", (y == 1).sum())
```

```python
"""
dùng KFold(n_splits=5, shuffle=True, random_state=7) (KHÔNG stratify),
với mỗi fold: huấn luyện trên train fold, dự đoán trên test fold,
in f1_score(average="macro", labels=[0,1], zero_division=0) và số mẫu lớp 1
có trong tập test của fold đó
"""
print("KFold (không stratify)")
kf = KFold(n_splits=5, shuffle=True, random_state=7)

for i, (train_idx, test_idx) in enumerate(kf.split(X, y), start=1):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    model = LogisticRegression(max_iter=1000, random_state=7)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    f1 = f1_score(y_test, y_pred, average="macro", labels=[0, 1], zero_division=0)
    so_luong_lop1 = (y_test == 1).sum()

    print(f"Fold {i}: macro-F1 = {f1:.4f} | số mẫu lớp 1 trong test = {so_luong_lop1}")
```

```python
# lặp lại với StratifiedKFold(n_splits=5, shuffle=True, random_state=7), so sánh
print("\nStratifiedKFold")
skf = StratifiedKFold(n_splits=4, shuffle=True, random_state=7)

for i, (train_idx, test_idx) in enumerate(skf.split(X, y), start=1):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    model = LogisticRegression(max_iter=1000, random_state=7)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    f1 = f1_score(y_test, y_pred, average="macro", labels=[0, 1], zero_division=0)
    so_luong_lop1 = (y_test == 1).sum()

    print(f"Fold {i}: macro-F1 = {f1:.4f} | số mẫu lớp 1 trong test = {so_luong_lop1}")
```
